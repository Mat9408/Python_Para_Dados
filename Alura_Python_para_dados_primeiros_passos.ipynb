IF - ELSE:

```python
idade_maria = int(input('Digite a idade de Maria: '))
idade_joao = int(input('Digite a idade de João: '))

if idade_maria > idade_joao:
    print('Maria é mais velha que João.')
elif idade_maria == idade_joao:
    print('Maria e João têm a mesma idade.')
else:
    print('João é mais velho que Maria.')
```

```python
nota_1 = float(input('Digite a nota 1: '))
nota_2 = float(input("Digite a nota 2: "))

media = (nota_1 + nota_2) / 2

if media >= 7:
    print(media)
    print("Aprovado")
elif media >= 5:
    print(media)
    print("Recuperação")
else:
    print(media)
    print("Reprovado")
```

```python
num_1 = int(input("Digite um número: "))
num_2 = int(input("Digite outro número: "))

if num_1 > num_2:
  print(f'O número {num_1} é maior que o número {num_2}.')
elif num_1 == num_2:
  print(f'Ambos são iguais.')
else:
  print(f'O número {num_2} é maior que o número {num_1}.')
```

```python
percentual_anterior = float(input("Digite o percentual de produção anterior: "))
percentual_atual = float(input("Digite o percentual de produção atual: "))
diferenca = percentual_atual - percentual_anterior

if diferenca > 0:
  print(f"Houve um aumento de {diferenca}% na produção.")
elif diferenca == 0:
  print(f"Não houve variação percentual de produção.")
else:
  print(f"Houve uma redução de {diferenca}% na produção.")
```
